```python
'''
Функция evalf()
Функция вычисляет данное числовое выражение с точностью до 100 цифр после плавающей точки. 
Она также принимает параметр subs, как объект словаря с числовыми значениями для символов.
'''

from sympy.abc import r 
expr = pi * r**2 
print(expr)
expr.evalf(100, subs = {r:5})
```

```python
from sympy import pi
expr = pi
print(expr)
expr.evalf(40)
```

```python
from sympy.abc import a,b
expr = a / b 
expr.evalf(80, subs={a: 100, b: 3})
```

```python
# Функция lambdify() переводит выражения SymPy в функции Python.
from sympy import *
x = Symbol('x')
expr = 1 / sin(x) 
f = lambdify(x, expr) 
f(3.14)
```

```python
from sympy.abc import a, b
expr = a**2 + b**2 
f = lambdify([a, b], expr) 
f(2, 3)
```

```python
from sympy.abc import a, b
import numpy as np
expr = a**2 + b**2 
f = lambdify([a, b], expr, "numpy") 
l1 = np.arange(1, 6) 
l2 = np.arange(6, 11) 
f(l1, l2)
```
